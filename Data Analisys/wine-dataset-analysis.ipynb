### Questão 6
Qual é a mediana da coluna `points` no DataFrame `reviews`?

```python
#@title Resposta Q6

def column_null_check(df, column_name):
  if df[column_name].isnull().values.any():
    print(f'Há valores nulos na coluna {column_name}.')
  else:
    print(f'Não há valores nulos na coluna {column_name}.')

column_null_check(wine_reviews, 'points')


median_points = wine_reviews['points'].median()
print(f'A mediana da coluna points é: {median_points}')
```

### Questão 10
Suponha que você seja um comprador assíduo de vinhos. Que vinho é considerado a melhor pechincha? Crie uma variável `bargain_wine` com o título de vinhos que apresentam a razão entre pontos-e-preço mais elevada no dataset.

```python
#@title Resposta Q10 - Razão

list_of_bargain_wines = []
for i in range(len(wine_reviews)):
  list_of_bargain_wines.append(wine_reviews.loc[i, 'points'] / wine_reviews.loc[i, 'price']) # Para cada linha, calcula a razão entre pontos e preço

wine_reviews['bargain_points_price'] = list_of_bargain_wines
bargain_wine = wine_reviews.loc[wine_reviews['bargain_points_price'] == wine_reviews['bargain_points_price'].max()] # Retorna apenas as linhas onde a razão é igual ao valor máximo
bargain_wine
```

```python
#@title Resposta Q10 - Mediana
# w = 0 # @param {"type":"number"}

price_median = wine_reviews['price'].median()
points_median = wine_reviews['points'].median()

print(f'\nConsiderando uma mediana de preço de {price_median:.2f} e uma mediana pontos de {points_median:.2f}, os vinho de maior barganha são:\n')

bargain_wine = wine_reviews.loc[(wine_reviews['price'] < price_median) & (wine_reviews['points'] > points_median)]
bargain_wine.sort_values(by='price', ascending=True)
```

### Questão 11
Há muita palavras que podemos usar para descrever uma garrafa de vinho. Um vinho é mais provável ser "tropical" ou "fruity"? Crie uma Series chamada `descriptor_counts` que armazene quantas vezes cada uma dessas palavras aparecem na coluna `description`. (Por razões de simplicidade, ignore as versões em maiúsculas dessas palavras).

```python
#@title Resposta Q11

tropical_count = wine_reviews['description'].map(lambda desc: 'tropical' in desc).sum()
fruity_count = wine_reviews['description'].map(lambda desc: 'fruity' in desc).sum()

descriptor_counts = pd.Series([tropical_count, fruity_count], index=['tropical', 'fruity'])
descriptor_counts
```

### Questão 12
Suponha que gostaríamos de traduzir o sistema atual de pontos para um sistema de notas mais simplificado. Um vinho com score de 95 ou mais receberia 3 estrelhas, um score de pelo menos 85 e não menos que 95 receberia 2 estrelas. Qualquer outro score de pontos recebe 1 estrela.

Crie uma séries `star_ratings` com o número de estrelas correspondendo a cada review no dataset.

```python
#@title Resposta Q12

def stars(row):
    if row.points >= 95:
        return 3
    elif row.points >= 85:
        return 2
    else:
        return 1

star_ratings = wine_reviews.apply(stars, axis='columns')
star_ratings
```

## Grouping and Sorting

### Questão 13
Crie uma `Series` cujo os índices é um revisor (`reviewers`) e cujo os valores é o score médio de revisão dado por aquele revisor. Dica: você precisa utlizar as colunas `taster_name` e `points`.

Em seguida, compare o score médio de revisão atribuído pelos vários revisores e o score médio global do dataset.

```python
#@title Resposta Q13

global_mean_points = wine_reviews['points'].mean()
print(f"\nO score médio global é: {global_mean_points:.2f}\n")

reviewer_mean_points = wine_reviews.groupby('taster_name')['points'].mean()
reviewer_mean_points = reviewer_mean_points.dropna()
reviewer_mean_points = reviewer_mean_points.sort_values(ascending=False)

reviewer_mean_points_diff = reviewer_mean_points - global_mean_points

reviewer_stats = pd.DataFrame({'Mean Points': reviewer_mean_points, 'Global Mean Diff.': reviewer_mean_points_diff})
reviewer_stats.reset_index(inplace=True)
reviewer_stats.rename(columns={'taster_name': 'Reviewer'}, inplace=True)
reviewer_stats
```
