### Questão 4

Crie dois tensores aleatórios de shape (2,3), carregue-os para GPU e execute a multiplicação de matrizes sobre eles na GPU. Lembre que você precisa habilitar o suporte da GPU no Colab. Defina uma semente aleatória como 1234 na GPU.

```python
#@title Resposta 004

torch.manual_seed(1234)
if torch.cuda.is_available():
  tensor1_gpu = torch.rand(2, 3)
  tensor2_gpu = torch.rand(2, 3).t()
  resultado_gpu = torch.matmul(tensor1_gpu, tensor2_gpu)
  print("\nResultado na GPU:\n", resultado_gpu)
else:
  print("GPU não disponível.")
```

### Questão 5

Encontre os valores máximo e mínimo da saída da questão 4.

```python
#@title Resposta 005

maximo = torch.max(resultado_gpu)
minimo = torch.min(resultado_gpu)
print("Máximo:", maximo.item())
print("Mínimo:", minimo.item())
```

### Questão 6
 Encontre os valores dos índices do máximo e mínimo da questão 4.

```python
#@title Resposta 006

indices_max = torch.argmax(resultado_gpu)
indices_min = torch.argmin(resultado_gpu)
print("Índice do máximo:", indices_max.item())
print("Índice do mínimo:", indices_min.item())
```

### Questão 8

* Crie três imagens sintéticas contendo 7 pixels de altura e 7 pixels de largura usando tensores aleatórios. Assuma que os valores de pixels aleatórios são inteiros e variam entre 0 a 255.
* Combine esses três tensores (imagens) em um único tensor maior de rank-3
* Reestruture este tensor de rank-3 de modo que seja adicionada mais uma dimensão de comprimento 1. Esta dimensão deve representar o canal de cor na escala de cinza. Espera-se que este tensor no final tenha o seguinte shape (#N_IMG, #CANAL, #ALTURA, #LARGURA)
* imprima os pixels da primeira imagem
* imprima os pixels da terceira imagem
* imprima os pixels no canal de cor da primeira imagem
* imprima os pixels no canal de cor da terceira imagem
* imprima os pixels das linhas ímpares no canal de cor da primeira imagem
* imprima os pixels das linhas ímpares no canal de cor da terceira imagem
* imprima os pixels das linhas pares no canal de cor da primeira imagem
* imprima os pixels das linhas ímpares no canal de cor da terceira imagem
* imprima os pixels das colunas ímpares no canal de cor da primeira imagem
* imprima os pixels das colunas pares no canal de cor da terceira imagem

```python
#@title Resposta 008

torch.manual_seed(42)
img1 = torch.randint(0, 256, (7, 7))
img2 = torch.randint(0, 256, (7, 7))
img3 = torch.randint(0, 256, (7, 7))

imagens = torch.stack([img1, img2, img3])
imagens = imagens.unsqueeze(1)

print("\nPrimeira imagem:\n", imagens[0, 0])
print("Terceira imagem:\n", imagens[2, 0])
print("Canal de cor primeira imagem:\n", imagens[0])
print("Canal de cor terceira imagem:\n", imagens[2])
print("Linhas ímpares primeira imagem:\n", imagens[0, 0, ::2])
print("Linhas ímpares terceira imagem:\n", imagens[2, 0, ::2])
print("Linhas pares primeira imagem:\n", imagens[0, 0, 1::2])
print("Linhas ímpares terceira imagem:\n", imagens[2, 0, ::2])
print("Colunas ímpares primeira imagem:\n", imagens[0, 0, :, ::2])
print("Colunas pares terceira imagem:\n", imagens[2, 0, :, 1::2])
```

### Questão 9

Crie um tensor que armazene as seguintes estatísticas dos pixels de cada imagem:
* soma das linhas
* soma das colunas
* moda
* média de todos
* desvio padrão

```python
#@title Resposta 009

estatisticas = {
    'soma_linhas': torch.sum(imagens, dim=3),
    'soma_colunas': torch.sum(imagens, dim=2),
    'moda': torch.mode(imagens.reshape(-1))[0],
    'media': torch.mean(imagens.double()),
    'desvio_padrao': torch.std(imagens.double())
}

print("Estatísticas:")
for k, v in estatisticas.items():
    print(f"{k}:\n{v}")
```

### Questão 10

Considere o tensor de imagens de rank-4 resultante da questão 8.
Realize a operação de `flatten` somente sobre as matrizes de pixels deste tensor e imprima o tensor resultante. Lembrando que flatten é uma operação sobre tensores que faz com que o shape de um tensor seja igual ao número de elementos contidos nele, ou seja, que transforme as imagens em um tensor de rank-1 (array 1-d).

```python
#@title Resposta 010

imagens_flat = imagens.reshape(imagens.shape[0], imagens.shape[1], -1)
print("\nTensor após flatten das matrizes de pixels:\n", imagens_flat)
print("\nShape após flatten:", imagens_flat.shape)
```
